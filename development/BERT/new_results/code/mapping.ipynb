```python
from fuzzywuzzy import fuzz
import re

def normalize_text(s: str) -> str:
    """统一文本格式：去换行、多空格，转小写"""
    if s is None:
        return ""
    s = re.sub(r'[\r\n]+', ' ', str(s))
    s = re.sub(r'\s{2,}', ' ', s)
    return s.strip().lower()

def is_similar(text1: str, text2: str, threshold: int = 90) -> bool:
    """
    判断两句话相似度是否 >= threshold
    返回布尔值 + 相似度分数
    """
    t1 = normalize_text(text1)
    t2 = normalize_text(text2)
    score = fuzz.partial_ratio(t1, t2)
    return score >= threshold, score

# 使用示例
s1 = "This spontaneous case was reported by a consumer."
s2 = "This spontaneous case was reported by a healthcare professional."

similar, score = is_similar(s1, s2, threshold=85)
print(f"相似度: {score}, 是否超过阈值: {similar}")
```

```python
import re

def normalize_format(text: str) -> str:
    """
    把带换行/空行的病例叙述压缩成单段文本
    1. 去掉换行、回车等控制符
    2. 多个空格合并成一个
    3. 去掉首尾空格
    """
    if text is None:
        return ""
    # 换行/回车 -> 空格
    text = re.sub(r"[\r\n]+", " ", text)
    # 多个空格 -> 一个
    text = re.sub(r"\s{2,}", " ", text)
    return text.strip()
isSame = []
for i in range(len(json_content)):
    
    # 示例
    text1 = subset["Narrative"].iloc[i]
    text2 = normalize_format(text1)
    print(text2)
    print(json_content[i])
    if is_similar(text2, json_content[i], threshold=90)[0]:
        isSame.append(True)
    else:
        isSame.append(False)
```
